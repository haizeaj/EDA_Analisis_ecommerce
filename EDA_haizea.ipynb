```python
# MÉTRICAS GLOBALES: TOTAL DE VISITAS Y EVENTOS
total_visitas = len(df)
total_eventos = df['eventos_interaccion'].sum()

print("--- RESUMEN GLOBAL ---")
print(f"Total de visitas (sesiones): {total_visitas}")
print(f"Total de eventos de interacción: {total_eventos}")
```

```python
# MÉTRICAS POR PAÍS: VISITAS, EVENTOS Y TRANSACCIONES
def tabla_por_pais(df):
    g = df.groupby("pais")
    t = pd.DataFrame({
        "visitas": g.size(),                              # Total de visitas/sesiones por país
        "total_eventos": g["eventos_interaccion"].sum(),  # Suma total de eventos de interacción
        "transacciones": g["transacciones"].sum(),        # Total de transacciones (conversiones)
    })
    return t.sort_values("visitas", ascending=False)

# Ejecutar la función
resumen_pais = tabla_por_pais(df)

print("--- RESUMEN POR PAÍS (VISITAS, EVENTOS Y TRANSACCIONES) ---")
print(resumen_pais)

# Destacados por categoría
print(f"\n-> País con más visitas:       {resumen_pais['visitas'].idxmax()} ({resumen_pais['visitas'].max()} visitas)")
print(f"-> País con más eventos:       {resumen_pais['total_eventos'].idxmax()} ({resumen_pais['total_eventos'].max()} eventos)")
print(f"-> País con más transacciones: {resumen_pais['transacciones'].idxmax()} ({resumen_pais['transacciones'].max()} transacciones)")
```

```python

# VISITAS Y EVENTOS POR CANAL
def tabla_visitas_y_eventos(df):
    g = df.groupby("canal")
    t = pd.DataFrame({
        "visitas_totales": g.size(),                            # Total de sesiones/visitas por canal
        "total_eventos": g["eventos_interaccion"].sum(),        # Suma total de eventos de interacción
        "promedio_eventos": g["eventos_interaccion"].mean(),    # Promedio de eventos por visita
    })
    return t.sort_values("visitas_totales", ascending=False)

# Ejecutar la función
resumen_canal = tabla_visitas_y_eventos(df)

print("RESUMEN: VISITAS Y EVENTOS POR CANAL")
print(resumen_canal.round(2))

# Conclusiones automáticas
print(f"\n-> Canal con más visitas totales: {resumen_canal['visitas_totales'].idxmax()}")
print(f"-> Canal con más eventos totales: {resumen_canal['total_eventos'].idxmax()}")
```

¿Qué canal genera más compras? Direct, tanto en tasa (2,35 %) como en número: 8.729 sesiones con compra, más de cuatro veces las de Organic Search (2.119). En ingresos también lidera, con 1,19 M$.

```python
# COMPARATIVA: RETENCIÓN, EVENTOS Y TRANSACCIONES POR CANAL
def tabla_retencion_y_conversion(df):
    g = df.groupby("canal")
    t = pd.DataFrame({
        "sesiones": g.size(),
        
        # 1. Métricas de Retención / Engagement
        "duracion_media_min": g["duracion_min"].mean(),      # Duración media de la visita en minutos
        "paginas_vistas_media": g["paginas_vistas"].mean(),  # Páginas vistas por sesión
        
        # 2. Métricas de Eventos
        "total_eventos": g["eventos_interaccion"].sum(),     # Suma total de eventos de interacción
        "promedio_eventos": g["eventos_interaccion"].mean(), # Eventos por sesión
        
        # 3. Métricas de Transacciones
        "total_transacciones": g["transacciones"].sum(),     # Total de conversiones/transacciones
    })
    
    # Ordenamos de mayor a menor según la duración media (retención principal)
    return t.sort_values("duracion_media_min", ascending=False)

# Ejecutar la función
resumen_canal = tabla_retencion_y_conversion(df)

print("--- COMPARATIVA POR CANAL: RETENCIÓN VS EVENTOS Y TRANSACCIONES ---")
print(resumen_canal.round(2))

# Destacados automáticos
print(f"\n-> Canal con mayor retención (duración media): {resumen_canal['duracion_media_min'].idxmax()}")
print(f"-> Canal con más páginas vistas por sesión:    {resumen_canal['paginas_vistas_media'].idxmax()}")
print(f"-> Canal con más eventos totales:              {resumen_canal['total_eventos'].idxmax()}")
print(f"-> Canal con más transacciones totales:        {resumen_canal['total_transacciones'].idxmax()}")
```

Hay un matiz: Google Analytics registra duración 0 en las sesiones de rebote. Si solo se cuentan las sesiones sin rebote, todos los canales quedan en torno a 4,7-5,1 minutos, salvo Social (1,9). ```python

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Cargar dataset
df = pd.read_csv(
    "data/Datos_corregidos_google.csv",
    low_memory=False
)

# 2. Crear variable de conversión
# Una sesión convierte si tiene al menos una transacción
df["converted"] = (df["transacciones"].fillna(0) > 0).astype(int)

# 3. Agrupar y calcular métricas por canal
channel_summary = (
    df.groupby("canal")
    .agg(
        total_sesiones=("visit_id", "count"),
        sesiones_convertidas=("converted", "sum"),
        tasa_conversion=("converted", "mean"),
        avg_paginas_vistas=("paginas_vistas", "mean"),
        duracion_media_seg=("duracion_seg", "mean")
    )
    .reset_index()
)

# Convertir tasa a porcentaje
channel_summary["tasa_conversion_pct"] = (
    channel_summary["tasa_conversion"] * 100
)

# Convertir duración a minutos
channel_summary["duracion_min"] = (
    channel_summary["duracion_media_seg"] / 60
)

# 4. Configuración estética
sns.set_theme(style="whitegrid", context="notebook")

fig, ax = plt.subplots(figsize=(12, 7), dpi=150)

# 5. Gráfico de burbujas
sns.scatterplot(
    data=channel_summary,
    x="avg_paginas_vistas",
    y="tasa_conversion_pct",
    hue="canal",
    size="duracion_min",
    sizes=(150, 850),
    palette="tab10",
    alpha=0.8,
    edgecolor="black",
    linewidth=0.7,
    ax=ax
)

# 6. Etiquetas de los canales
for _, row in channel_summary.iterrows():
    ax.annotate(
        row["canal"],
        (
            row["avg_paginas_vistas"],
            row["tasa_conversion_pct"]
        ),
        xytext=(7, 7),
        textcoords="offset points",
        fontsize=9,
        fontweight="bold"
    )

# 7. Títulos y ejes
ax.set_title(
    "Comportamiento de navegación y tasa de conversión por canal",
    fontsize=14,
    fontweight="bold",
    pad=18
)

ax.set_xlabel(
    "Promedio de páginas vistas por sesión",
    fontsize=11
)

ax.set_ylabel(
    "Tasa de conversión (%)",
    fontsize=11
)

# 8. Leyenda
ax.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    title="Canal y duración media",
    frameon=True
)

sns.despine()
plt.tight_layout()
plt.show()

# 9. Mostrar tabla de resultados
print(
    channel_summary[
        [
            "canal",
            "total_sesiones",
            "sesiones_convertidas",
            "tasa_conversion_pct",
            "avg_paginas_vistas",
            "duracion_min"
        ]
    ].round(2).to_string(index=False)
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Cargar los datos
df = pd.read_csv("data/Datos_corregidos_google.csv")

# 2. Crear variable de conversión
# 1 = sesión con compra
# 0 = sesión sin compra
df["conversion"] = (df["transacciones"] > 0).astype(int)

# 3. Calcular tasa de conversión por canal
conversion = df.groupby("canal")["conversion"].mean() * 100

# 4. Ordenar de mayor a menor
conversion = conversion.sort_values(ascending=False)

# 5. Crear gráfica
plt.figure(figsize=(10, 6))

plt.bar(conversion.index, conversion.values)

# 6. Personalizar gráfica
plt.title("Tasa de conversión por canal de adquisición")
plt.xlabel("Canal de adquisición")
plt.ylabel("Tasa de conversión (%)")

plt.xticks(rotation=30)
plt.grid(axis="y", alpha=0.3)

# 7. Mostrar porcentajes encima de las barras
for i, valor in enumerate(conversion.values):
    plt.text(i, valor, f"{valor:.2f}%",
             ha="center", va="bottom")

# 8. Mostrar gráfica
plt.tight_layout()
plt.show()

```

Primera interpretación:
- Direct presenta la tasa de conversión más alta, con un 2,35 %.
- Display y Paid Search también muestran tasas relativamente elevadas.
- Organic Search tiene mucho volumen de tráfico, pero su tasa es inferior a la de Direct.
- Social genera un volumen considerable de sesiones, pero convierte muy poco.
Esto demuestra por qué es importante analizar las tasas y no únicamente el volumen de conversiones.

3- HIPOTESIS -> Los canales con mayor volumen de tráfico no son necesariamente los que generan más ingresos por sesión.
Variables: canal, sesiones, ingresos y transacciones.

```python
# Cargar CSV
df = pd.read_csv("data/Datos_corregidos_google.csv")

# Eliminar sesiones duplicadas
df = df.drop_duplicates(
    subset=["visitor_id", "visit_id"]
)

# Agrupar por canal
resultado = df.groupby("canal").agg(
    sesiones=("visit_id", "count"),
    transacciones=("transacciones", "sum"),
    ingresos=("ingresos", "sum")
).reset_index()

# Ordenar por volumen de sesiones
resultado = resultado.sort_values(
    by="sesiones", ascending=False
)

# Gráfico
plt.figure(figsize=(10, 6))

plt.bar(
    resultado["canal"],
    resultado["sesiones"]
)

plt.title("Volumen de sesiones por canal de adquisición")
plt.xlabel("Canal de adquisición")
plt.ylabel("Número de sesiones")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import pandas as pd
import seaborn as sns

# 1. Cargar datos
df = pd.read_csv("data/Datos_corregidos_google.csv")

# 2. Definir la lógica de compra y la función de conversión
df["compra"] = (df["transacciones"] > 0).astype(int)


def tabla_conversion(df):
  g = df.groupby("canal")
  t = pd.DataFrame({
      "sesiones": g.size(),
      "sesiones_con_compra": g["compra"].sum(),
      "tasa_conversion_pct": g["compra"].mean() * 100,
      "transacciones": g["transacciones"].sum(),
  })
  return t.sort_values("tasa_conversion_pct", ascending=False)


# 3. Calcular y mostrar resultados por consola
conversion = tabla_conversion(df)
print("4. CONVERSIÓN POR CANAL")
print(conversion)
print(
    f"\n-> Canal con mayor tasa de conversión:        {conversion['tasa_conversion_pct'].idxmax()}"
)
print(
    f"-> Canal que genera más compras (en número): {conversion['sesiones_con_compra'].idxmax()}"
)
print(f"Conversión global: {df['compra'].mean() * 100:.2f}%")

# 4. Preparar los datos para la visualización (reseteando índice)
# Ordenamos por 'sesiones' de forma ascendente para que el gráfico de barras horizontales quede ordenado visualmente
resultado = conversion.reset_index().sort_values("sesiones", ascending=True)

# 5. Crear los dos gráficos comparativos
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Gráfico 1: Volumen de tráfico (Sesiones)
sns.barplot(data=resultado, x="sesiones", y="canal", ax=axes[0], palette="Blues_r")
axes[0].set_title("Volumen de tráfico por canal", fontsize=12, fontweight="bold")
axes[0].set_xlabel("Número de sesiones", fontsize=10)
axes[0].set_ylabel("Canal de adquisición", fontsize=10)
axes[0].xaxis.set_major_formatter(
    ticker.StrMethodFormatter("{x:,.0f}")
)  # Separador de miles

# Gráfico 2: Tasa de conversión (%)
sns.barplot(
    data=resultado,
    x="tasa_conversion_pct",
    y="canal",
    ax=axes[1],
    palette="Greens_r",
)
axes[1].set_title(
    "Tasa de conversión por canal", fontsize=12, fontweight="bold"
)
axes[1].set_xlabel("Tasa de conversión (%)", fontsize=10)
axes[1].set_ylabel("")
axes[1].xaxis.set_major_formatter(
    ticker.PercentFormatter(xmax=100, decimals=1)
)  # Formato porcentaje automático

# Ajustar diseño general
fig.suptitle(
    "Volumen de tráfico y eficiencia transaccional por canal",
    fontsize=14,
    fontweight="bold",
    y=1.02,
)

plt.tight_layout()
plt.show()
```

H3: Los canales con mayor volumen de tráfico no son necesariamente los que generan más transacciones por sesión.

Los resultados muestran diferencias importantes:
- Direct es el canal con más sesiones (371.469) y también presenta una de las mayores ratios de transacciones por sesión (0,0247).
- Display, con solo 6.182 sesiones, consigue una ratio muy similar (0,0243).
- Social, pese a registrar 226.117 sesiones, presenta una ratio muy baja (0,0006).
- Paid Search, con 13.079 sesiones, alcanza 0,019 transacciones por sesión.
Hipótesis respaldada parcialmente
Conclusión para tu EDA:

El volumen de tráfico no garantiza una mayor generación de transacciones por sesión. Aunque Direct concentra tanto el mayor tráfico como una de las ratios más elevadas, Display consigue una eficiencia transaccional prácticamente equivalente con muchas menos sesiones, mientras que Social presenta un volumen elevado de visitas y una ratio muy inferior. Por tanto, los resultados respaldan que existen diferencias en la capacidad de los canales para generar transacciones en relación con el tráfico que reciben.

Mi matiz: no diría que los canales con menos tráfico siempre son mejores. Direct sigue destacando tanto en volumen como en transacciones por sesión. Lo interesante de tu análisis es demostrar que el volumen de visitas, por sí solo, no permite evaluar la eficiencia de un canal.

HIPOTESIS PRINCIPAL - El canal de adquisición y el comportamiento de navegación de los usuarios están asociados a diferencias en la tasa de conversión de un e-commerce.
